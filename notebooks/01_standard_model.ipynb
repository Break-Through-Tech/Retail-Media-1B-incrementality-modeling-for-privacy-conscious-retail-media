## Steps

### 1. Locate And Load The Prepared Data

The path logic works when the notebook is run from the repository root or from the
`notebooks` folder.

````python
if Path("data").is_dir():
    DATA_DIR = Path("data")
elif Path("../data").is_dir():
    DATA_DIR = Path("../data")
else:
    raise FileNotFoundError(
        "Could not find the data folder. Clone the repository and run the notebook "
        "from the repository root or notebooks folder."
    )

split_paths = {
    "train": DATA_DIR / "train.csv.gz",
    "validation": DATA_DIR / "validation.csv.gz",
    "test": DATA_DIR / "test.csv.gz",
}

missing_files = [str(path) for path in split_paths.values() if not path.exists()]
if missing_files:
    raise FileNotFoundError(f"Missing prepared files: {missing_files}")

splits = {name: pd.read_csv(path) for name, path in split_paths.items()}
train = splits["train"]
validation = splits["validation"]
test = splits["test"]

pd.DataFrame(
    {
        "split": list(splits),
        "rows": [len(frame) for frame in splits.values()],
        "columns": [frame.shape[1] for frame in splits.values()],
    }
)
````

### 3. Review Treatment And Outcome Rates

`treatment` is randomized assignment and defines the primary treatment and control groups.
`visit` is the primary outcome. `conversion` is an optional secondary outcome because it is
much rarer. `exposure` happens after assignment and must not be used as a normal audience
feature or as a filter for the primary analysis.

````python
rate_rows = []
for name, frame in splits.items():
    rate_rows.append(
        {
            "split": name,
            "rows": len(frame),
            "treatment_rate": frame["treatment"].mean(),
            "visit_rate": frame["visit"].mean(),
            "conversion_rate": frame["conversion"].mean(),
            "exposure_rate": frame["exposure"].mean(),
        }
    )

split_rates = pd.DataFrame(rate_rows).set_index("split")
split_rates.round(6)
````

````python
treatment_summary = (
    train.groupby("treatment")["visit"]
    .agg(records="size", visits="sum", visit_rate="mean")
    .rename(index={0: "control", 1: "treatment"})
)
treatment_summary
````

````python
intention_to_treat_lift = (
    treatment_summary.loc["treatment", "visit_rate"]
    - treatment_summary.loc["control", "visit_rate"]
)
print(f"Overall training-split visit uplift: {intention_to_treat_lift:.3%}")

ax = treatment_summary["visit_rate"].plot(
    kind="bar",
    color=["#637083", "#247BA0"],
    title="Training Visit Rate By Randomized Assignment",
    ylabel="Visit rate",
    xlabel="Assignment group",
    rot=0,
)
ax.yaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1.0))
plt.tight_layout()
plt.show()
````

### 5. Prepare Modeling Inputs

Only `f0` through `f11` are eligible audience model features. Keep treatment assignment and
outcomes separate. Do not use `exposure` as a model feature.

````python
X_train = train[FEATURE_COLUMNS].copy()
treatment_train = train["treatment"].copy()
visit_train = train["visit"].copy()

treatment_probability = treatment_train.mean()
transformed_visit = visit_train * (
    (treatment_train - treatment_probability)
    / (treatment_probability * (1 - treatment_probability))
)

print(f"Training treatment probability: {treatment_probability:.4f}")
print(f"Modeling matrix shape: {X_train.shape}")
print(f"Transformed-outcome mean: {transformed_visit.mean():.6f}")
````

## Logistic Regression Model 

- Built as a baseline benchmark model trained on treatment records to predict the probability of a visit under treatment

````python
# Create labeled examples from treatment-assigned records only.
# The treatment assignment is randomized; exposure is post-assignment and is excluded.
treatment_mask = train["treatment"].eq(1)
X_response_train = train.loc[treatment_mask, FEATURE_COLUMNS]
y_response_train = train.loc[treatment_mask, "visit"]

print(f"Treatment records used for training: {len(X_response_train):,}")
print(f"Treatment visit rate: {y_response_train.mean():.3%}")
print(f"Feature matrix shape: {X_response_train.shape}")
````

### Fit And Evaluate The Response Model

This model estimates $P(visit=1 \mid X, treatment=1)$: the probability that a treatment-assigned record visits, given its audience features. It is a response model, not an uplift model, so it does not estimate the incremental effect of advertising.

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import brier_score_loss, log_loss, roc_auc_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

response_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1_000, random_state=RANDOM_SEED),
)
response_model.fit(X_response_train, y_response_train)

validation_treatment = validation["treatment"].eq(1)
X_response_validation = validation.loc[validation_treatment, FEATURE_COLUMNS]
y_response_validation = validation.loc[validation_treatment, "visit"]
validation_probability = response_model.predict_proba(X_response_validation)[:, 1]

response_metrics = pd.Series(
    {
        "validation_rows": len(y_response_validation),
        "validation_visit_rate": y_response_validation.mean(),
        "roc_auc": roc_auc_score(y_response_validation, validation_probability),
        "log_loss": log_loss(y_response_validation, validation_probability),
        "brier_score": brier_score_loss(y_response_validation, validation_probability),
    }
)
response_metrics
````

````python
# Rank validation treatment records by predicted visit probability.
# rank(method="first") makes ties deterministic before forming deciles.
validation_ranked = validation.loc[validation_treatment, ["treatment", "visit"]].copy()
validation_ranked["predicted_visit_probability"] = validation_probability
validation_ranked["rank"] = validation_ranked["predicted_visit_probability"].rank(
    method="first", ascending=False
)
validation_ranked["decile"] = pd.qcut(
    validation_ranked["rank"], q=10, labels=False
) + 1

response_deciles = (
    validation_ranked.groupby("decile", observed=True)
    .agg(
        records=("visit", "size"),
        visits=("visit", "sum"),
        observed_visit_rate=("visit", "mean"),
        mean_predicted_probability=("predicted_visit_probability", "mean"),
    )
    .sort_index()
)
response_deciles["lift_vs_validation_treatment_average"] = (
    response_deciles["observed_visit_rate"] / y_response_validation.mean()
)
response_deciles
````
